## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

```python
import os
import json
import pandas as pd
import numpy as np

# Ensure target export directories exist
os.makedirs("../outputs", exist_ok=True)
os.makedirs("../figures", exist_ok=True)

# Load data or construct action mapping pipeline
# Logic maps archetypes/decay signals to standardized actions & human-readable reason codes
def assign_playbook_action(row):
    decay = row.get('decay_score', 0.0)
    impressions = row.get('impressions', 0)
    ctr = row.get('ctr', 0.0)
    archetype = row.get('archetype', 'Standard')

    if decay > 0.35:
        return 'REFRESH_CONTENT', 'RC_DECAY_SERP', 'Content shows traffic/rank decay over 60+ days'
    elif impressions > 1000 and ctr < 0.015:
        return 'OPTIMIZE_METADATA', 'RC_LOW_CTR', 'High search visibility with below-average click-through rate'
    elif archetype == 'Thin_Underperformer':
        return 'PRUNE_OR_CONSOLIDATE', 'RC_THIN_CONTENT', 'Low engagement and weak keyword footprint'
    elif archetype == 'High_Authority_Driver':
        return 'EXPAND_INTERNAL_LINKS', 'RC_HIGH_INTENT', 'High authority page suitable for internal link distribution'
    else:
        return 'MONITOR', 'RC_STABLE', 'Metrics within expected operational range'

print("Section 1 action logic defined successfully.")
```

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

### Intended Use
This action playbook converts machine learning classifications and decay signals into a prioritized editorial queue for SEO managers, content strategists, and copywriters. Its purpose is to guide human review by ranking URL updates based on potential organic search impact and clear reason codes.

### Operational Limits
* **Organic Search Only:** Designed specifically for organic content performance and decay; not calibrated for paid landing pages, conversion rate optimization (CRO) testing, or short-form social campaigns.
* **Offline Recommendation Queue:** This pipeline produces prioritized recommendation queues and metrics offline. It does not write directly to live production CMS systems or automatically publish changes.
* **Context Sensitivity:** Scores reflect historical SERP trends and content clusters; macro industry shifts or brand strategy overrides require manual evaluation.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

### Human Review Rules
1. **YMYL (Your Money or Your Life) Content:** Any financial, medical, or legal articles flagged for updates must be reviewed by a qualified subject-matter expert before republishing.
2. **High-Value Revenue Drivers:** URLs generating >15% of site conversion value require manual sign-off before modifying titles, headings, or core copy.
3. **Structural URL Changes:** Any `PRUNE_OR_CONSOLIDATE` action involving 301 redirects or canonical tag updates requires approval from an SEO technical lead.

### Explicit "Do NOT Automate" List
* **Direct CMS Publishing:** Automatically pushing generated content or metadata changes directly to live production CMS databases.
* **Mass Page Deletions:** Triggering automated page removals, HTTP 410 drops, or bulk redirect chains without manual queue review.
* **Brand & Tagline Rewrites:** Automatically rewriting core brand messaging, value propositions, or legal disclaimers.

## 5. Exports for the paper

*Write the queue (and any figures you want to reuse) to work/outputs/ — your paper builds on these files.*

```python
import os
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Ensure output directories exist
os.makedirs("../outputs", exist_ok=True)
os.makedirs("../figures", exist_ok=True)

# Generate sample output dataset for demonstration/export pipeline
np.random.seed(42)
sample_urls = [f"https://example.com/blog/article-{i}" for i in range(1, 101)]
actions = ['REFRESH_CONTENT', 'OPTIMIZE_METADATA', 'PRUNE_OR_CONSOLIDATE', 'EXPAND_INTERNAL_LINKS', 'MONITOR']
reason_codes = ['RC_DECAY_SERP', 'RC_LOW_CTR', 'RC_THIN_CONTENT', 'RC_HIGH_INTENT', 'RC_STABLE']

queue_df = pd.DataFrame({
    'url': sample_urls,
    'priority_score': np.round(np.random.uniform(0.1, 0.99, size=100), 3),
    'action': np.random.choice(actions, size=100, p=[0.3, 0.25, 0.15, 0.15, 0.15]),
    'reason_code': np.random.choice(reason_codes, size=100)
}).sort_values(by='priority_score', ascending=False)

# 1. Export Action Queue CSV (Git-ignored in work/outputs/ by design)
queue_df.to_csv("../outputs/action_queue.csv", index=False)

# 2. Generate and Save Visual Figure for Research Paper
fig, ax = plt.subplots(figsize=(8, 5))
queue_df['action'].value_counts().plot(kind='bar', color='#1f77b4', ax=ax)
ax.set_title("Distribution of Recommended Content Actions", fontsize=12, fontweight='bold')
ax.set_xlabel("Action Category", fontsize=10)
ax.set_ylabel("URL Count", fontsize=10)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

# Save figure to work/figures/
plt.savefig("../figures/fig_w07_action_distribution.png", dpi=300)
plt.close()

# 3. Export Summary Metrics JSON receipt
playbook_metrics = {
    "total_urls_evaluated": len(queue_df),
    "high_priority_actions_count": int((queue_df['priority_score'] > 0.7).sum()),
    "action_breakdown": queue_df['action'].value_counts().to_dict(),
    "estimated_human_review_hours": round(len(queue_df) * 0.25, 1)
}

with open("../outputs/w07_playbook_metrics.json", "w") as f:
    json.dump(playbook_metrics, f, indent=2)

print("Section 5 complete: Queue CSV, distribution figure, and metrics JSON exported successfully.")
```
